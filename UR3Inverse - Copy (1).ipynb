```python

import numpy as np
import sympy as sy
from sympy import Inverse
import math

q1, q2, q3, q4, q5, q6 = sy.symbols('q1 q2 q3 q4 q5 q6')


A1 = sy.Matrix([[sy.cos(q1), 0, sy.sin(q1), 0],
    [sy.sin(q1), 0, -sy.cos(q1), 0],
    [0, 1, 0, 0.1519],
    [0, 0, 0, 1]])
A2 = sy.Matrix([[sy.cos(q2), -sy.sin(q2), 0, -0.24365*sy.cos(q2)],
    [sy.sin(q2),  sy.cos(q2), 0, -0.24365*sy.sin(q2)],
    [      0,        0, 1,                     0],
    [      0,        0, 0,                     1]])
A3 = sy.Matrix([[sy.cos(q3), -sy.sin(q3), 0, -0.2132*sy.cos(q3)],
    [sy.sin(q3),  sy.cos(q3), 0,-0.2132*sy.sin(q3)],
    [      0,        0, 1,                   0],
    [      0,        0, 0,                   1]])

A4 = sy.Matrix([[sy.cos(q4), 0,  sy.sin(q4),          0],
    [sy.sin(q4), 0, -sy.cos(q4),          0],
    [      0, 1,        0, 0.11235],
    [      0, 0,        0,          1]])
A5 = sy.Matrix([[sy.cos(q5),  0, -sy.sin(q5),          0],
    [sy.sin(q5),  0,  sy.cos(q5),          0],
    [      0, -1,        0, 0.08535],
    [      0,  0,        0,          1]])

A6 = sy.Matrix([[sy.cos(q6), -sy.sin(q6), 0,         0],
    [sy.sin(q6),  sy.cos(q6), 0,         0],
    [      0,        0, 1, 0.0819],
    [      0,        0, 0,         1]])





P60 = np.array([-0.08175,0.14151,0.41641])

R60 = np.array([[-0.59879,-0.80064,0.02052],
                [0.71189,-0.54380,-0.44441],
                [0.36697,-0.25150,0.89559]])


P60_vector = sy.Matrix(P60)

T60 = sy.Matrix([[R60[0, 0], R60[0, 1], R60[0, 2], P60_vector[0]],
                 [R60[1, 0], R60[1, 1], R60[1, 2], P60_vector[1]],
                 [R60[2, 0], R60[2, 1], R60[2, 2], P60_vector[2]],
                 [0, 0, 0, 1]])


P50 = P60 - 0.0819 * R60[:, 2]

x5 = P50[0]
y5 = P50[1]
z5 = P50[2]


psi = np.arctan2(y5, x5)
phi = -np.arccos(0.1124 / np.sqrt(x5 ** 2 + y5 ** 2)) #oder np.arccos

theta1 = psi + phi + np.pi / 2


P61z = P60[0] * np.sin(theta1) - P60[1] * np.cos(theta1)
theta5 = np.arccos((P61z - 0.1124) / 0.0819) # oder -np.arccos




A1d = A1.replace(q1,theta1)
T16 = sy.Inverse(A1d)@T60
T16F= sy.Inverse(T16)

theta6 = sy.atan2(-T16F[1, 2] / sy.sin(theta5), T16F[0, 2] / sy.sin(theta5))




v1 = np.array([0, -0.1124, 0, 1])
v2 = np.array([0, 0, 0, 1])



A5d = A5.replace(q5,theta5)
A6d = A6.replace(q6,theta6)


T40 = T60@(sy.Inverse(A5d@ A6d))
P30 = T40 @ v1 - v2

x30 = P30[0]
y30 = P30[1]
z30 = P30[2]

N = (x30 ** 2 + y30 ** 2 + ((z30 - 0.1519) ** 2) - 0.10481) / 0.103917
theta3 = sy.acos(N) #oder -sy.acos



s=z30-0.1519
r=sy.sqrt(x30**2+y30**2)
th2= sy.asin(((-0.2437-0.2133*sy.cos(theta3))*s+0.2133*sy.sin(theta3)*r)/(r**2+s**2))
th22=math.degrees(sy.pi+th2)
theta2=math.radians(th22)



T3=A1*A2*A3
T30 = T3.replace(q1,theta1).replace(q2,theta2).replace(q3,theta3)
T63 = sy.Inverse(T30)@T60

th4 = sy.atan2(-T63[1, 2], -T63[0, 2])

th44=math.degrees(th4)
theta4=math.radians(th44)



print("theta1:", theta1)
print("theta2:", theta2)
print("theta3:", theta3)
print("theta4:", theta4)
print("theta5:", theta5)
print("theta6:", theta6)

A4d = A4.replace(q4,theta4)
A3d = A3.replace(q3,theta3)
A2d = A2.replace(q2,theta2)
F60=A1d@A2d@A3d@A4d@A5d@A6d

print("")
sy.pprint(F60.evalf(5))
print("")

sy.pprint(T60)
```
